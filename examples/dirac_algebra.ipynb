## Momenta, indices and a trace

tr(p̸ γ^μ q̸ γ^ν)

```python
p, q, k = momenta("p q k")
mu, nu, rho, sig = lorentz_indices("mu nu rho sig")
m = var("m")
```

```python
expr = slash(p) * gamma(mu) * slash(q) * gamma(nu)
show(expr)
```

```python
show(dirac_trace(expr))
```

With masses. A propagator numerator is written as in the book.

```python
show(dirac_trace((slash(p) + m) * gamma(mu) * (slash(q) + m) * gamma(nu)))
```

## gamma5

γ^5 = iγ^0γ^1γ^2γ^3, so tr(γ^μγ^νγ^ργ^σγ^5) = -4iε^{μνρσ}.

```python
show(dirac_trace(gamma(mu) * gamma(nu) * gamma(rho) * gamma(sig) * gamma5()))
```

## Contractions and kinematics

```python
show(contract(gamma(mu) * gamma(nu) * gamma(mu)))
```

```python
d = var('d')
show(contract(gamma(mu) * gamma(nu) * gamma(mu), dim=d))
```

```python
s = var('s')
r = dirac_trace(slash(p) * slash(q) * slash(p) * slash(q))
show(r)
show(r.subs({dot(p, p): m^2, dot(q, q): 0, dot(p, q): s/2}))
```

## e+ e- → μ+ μ-

M = v̄(p2) γ^μ u(p1) ū(k1) γ_μ v(k2), electrons massless.

```python
p1, p2, k1, k2 = momenta("p1 p2 k1 k2")
M_mu = var("M_mu")
M = vbar(p2) * gamma(mu) * u(p1) * ubar(k1, M_mu) * gamma(mu) * v(k2, M_mu)
show(M)
```

```python
show(conjugate(M))
```

```python
show(spin_sum(M * conjugate(M)) / 4)
```

This is Peskin and Schroeder eq. (5.10) without e^4/q^4, with p = p1, p' = p2, k = k1, k' = k2.
